# Flow law conversion
This script shows how to convert laboratory-derived flow laws expressed in the "standard" form to the effective viscosity formulation dependent on the invariants

From laboratory experiments, one retrieves an experimentaly parametrised relation between strain-rate ($\dot{\epsilon}$) and stress ($\sigma$) for a specific configuration (experimental aparatus):

```python
flab = eps - C * sig**n * d**m # Standard form as from lab experiment
display(Eq(eps, solve(flab, eps)[0]))
```

where $C$ is a proportionnality coefficient, $n$ is the power-law exponent and $d$ the grain size.

In order to utilise this relation in numerical models (and multiple spatial dimensions) one needs to reformulate it with an effective viscosity ($\eta = \tau_\mathrm{II} / (2 \dot{\epsilon}_{\mathrm{II}})$) function of the second strain-rate or stress invariants ($\dot{\epsilon}_{\mathrm{II}}$ or $\tau_\mathrm{II}$).

In this framework, the strain-rate -- stress relation is formulated as:

```python
finv = flab.subs(eps, 2/sqrt(3)*eII ).subs(sig, sqrt(3)*tII)
display(Eq(eps, solve(finv, eII)[0])) # Introduce invariants
```

One can then solve for the effective viscosity as function of the strain-rate invariant $\dot{\epsilon}_{\mathrm{II}}$:

```python
f1 = finv.subs(tII, 2*eta*eII) # Solve for viscosity as function strain rate invariant
display(Eq(eta, nsimplify(solve(f1, eta)[0])) )
```

or solve for the effective viscosity as function of the stress invariant $\tau_\mathrm{II}$:

```python
f2 = finv.subs(eII, tII/2/eta) # Solve for viscosity as function stress invariant
display(Eq(eta, simplify(expand(solve(f2, eta)[0]))) )
```

For linear ($n=1$) creep with grain size dependence, one gets:

```python
flin = f2.subs(n, 1.0) # For linear creep, set n = 1
display( Eq(eta_dif, nsimplify(solve(flin, eta)[0])) )
```
